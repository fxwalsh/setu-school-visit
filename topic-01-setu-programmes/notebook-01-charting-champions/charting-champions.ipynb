## 1. Make a one-bar chart

These medal values are a short historical teaching sample from the original project. They are **not current Olympic totals**.

```python
us = ["United States", 2399]

chart = pygal.Bar()
chart.title = "Olympic medals - historical teaching sample"
chart.add(us[0], us[1])
display(SVG(chart.render()))
```

The list `us` keeps a country name and its count together. Python indexes start at zero: `us[0]` is the label and `us[1]` is the value.

## 2. Add more countries with lists

Add the remaining teaching-sample values. Each list has a name at index 0 and a count at index 1. The notebook output appears directly below the cell.

```python
teams = [
    ["United States", 2399],
    ["Russia", 1413],
    ["Great Britain", 1304],
    ["France", 780],
    ["Germany", 671],
]

chart = pygal.Bar()
chart.title = "Olympic medals - historical teaching sample"
for team in teams:
    chart.add(team[0], team[1])
display(SVG(chart.render()))
```

## 3. Read CSV rows

The original lab reads a CSV file from disk. To keep this Tutors notebook self-contained in the browser, the same CSV rows are included as text and parsed with Python's built-in `csv` module. This also handles quoted commas more safely than splitting each line manually.

```python
medals_csv = """United States,2399
Russia,1413
Great Britain,1304
France,780
Germany,671
"""

medal_rows = list(csv.reader(StringIO(medals_csv)))
print(medal_rows)
```

Each CSV value starts as text. Convert the second column to a whole number with `int()` before adding it to the chart.

```python
chart = pygal.Bar()
chart.title = "Olympic medals - read from CSV"
for team, raw_count in medal_rows:
    chart.add(team, int(raw_count))
display(SVG(chart.render()))
```

## 4. Compare fictional datasets

These three countries and every value below are fictional practice data. Population is in millions; GDP is in billions of example currency units. Use `float()` for decimal values and put units in every chart title.

```python
practice_data = [
    ("Fictional medals (count)", "Exampleland,12\nSample Isles,20\nDemovia,8\n", int),
    ("Fictional population (millions)", "Exampleland,5.2\nSample Isles,1.8\nDemovia,12.4\n", float),
    ("Fictional GDP (billions of example currency)", "Exampleland,150.5\nSample Isles,80.2\nDemovia,320.8\n", float),
]

for title, csv_text, convert in practice_data:
    chart = pygal.Bar()
    chart.title = title
    for country, raw_value in csv.reader(StringIO(csv_text)):
        chart.add(country, convert(raw_value))
    display(SVG(chart.render()))
```

**Discuss:** Does the fictional country with the largest population or GDP also have the most medals? Why can't invented data support a claim about real countries? What other information might matter, such as years participating, sports investment, or events entered?

## 5. Remix: make a class chart

Use anonymous class totals only; do not collect names. These are example counts. Replace them with your class's totals, or try a pie chart to compare how each display communicates the same data.

```python
survey_csv = """Football,8
Surfing,5
Cycling,6
Other,4
"""

pie = pygal.Pie()
pie.title = "Example activity preferences (anonymous counts)"
for activity, raw_count in csv.reader(StringIO(survey_csv)):
    pie.add(activity, int(raw_count))
display(SVG(pie.render()))
```
